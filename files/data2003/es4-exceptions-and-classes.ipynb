**An aside on `raise`.** Python raises an exception on its own when something goes wrong (`int("abc")` raises `ValueError`). Your own code can raise one on purpose, too, when it is handed something it should refuse. A method that raises stops at that line, so nothing after it runs and the object is left as it was. The code that called the method then decides what to do:

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError(f"Cannot withdraw {amount}: the balance is only {self.balance}")
        self.balance -= amount

acct = Account("Ana", 50)
acct.withdraw(20)
print(acct.balance)            # 30

try:
    acct.withdraw(100)
except ValueError as error:
    print("Refused:", error)   # Refused: Cannot withdraw 100: the balance is only 30
print(acct.balance)            # still 30
```

Why raise instead of printing a message? A printed message is easy to miss, and the code that called the method carries on as if the withdrawal worked. A raised error cannot be ignored: the caller either catches it, as above, or the program stops and says why. Exercise 5 asks you to do this.
